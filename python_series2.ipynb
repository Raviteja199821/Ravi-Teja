Sorting Numpy Array

```python
arr = np.array([2,1,5,3,7])
sorted_arr = np.sort(arr)
print(sorted_arr)
```

```python
np.shares_memory(arr, sorted_arr)
```

```python
#Inplace Sorting

arr.sort()
print(arr)
```

```python
np.shares_memory(arr,sorted_arr)
```

```python
#Sorting in descending order
arr = np.array([2,1,5,3,7])
sorted_arr_desc = np.sort(arr)[:: -1]
print(sorted_arr_desc)
```

```python
arr_2d = np.array([[5,2,8], [1,6,7], [4,3,9]])
sorted_arr_2d_col = np.sort(arr_2d, axis=0)
print(sorted_arr_2d_col)
```

```python
sorted_arr_2d_row = np.sort(arr_2d, axis =1)
print(sorted_arr_2d_row)
```

```python
#Sorting with indexes
arr = np.array([2,1,3,5,7])
sort_indices = np.argsort(arr)
print(sort_indices)
print(arr[sort_indices])
```

```python
# Lex Sort

a = np.array([1,5,1,4,3,4,4]) #First Key
b = np.array([9,4,0,4,0,2,1]) #Second key
sorted_index = np.lexsort((b,a))
print(sorted_index)

print(a[sorted_index],b[sorted_index])
```

Numpy Reshaping

```python
arr = np.arange(6)
reshaped_arr = arr.reshape((2,3))
print("reshaped_array:\n", reshaped_arr)
```

```python
#Using flatten
flat_arr = reshaped_arr.flatten()
print("Flattened array (flatten):", flat_arr)
```

```python
#Using ravel
ravel_arr = reshaped_arr.ravel()
print("Flattened array (ravel):", ravel_arr)
```

Dataset Manipulation Using Numpy

```python
# define the dtype for each column to use when reading the file
dtype = [('ID', 'i4'), ('FirstName', 'U10'), ('LastName', 'U10'),
         ('Gender', 'U10'), ('Age', 'i4'), ('City', 'U15')]
```

```python
data = np.genfromtxt('/sample_people.txt', delimiter=',', skip_header=1,
                     usecols=(0, 1, 2, 3, 4, 8), dtype=dtype, encoding='utf-8')
```

```python
data
```

```python
#Accessing specific fields
ages = data['Age']
print('Ages:', ages[:5])
```

```python
#Filtering data based on a condition
young_people = data[data['Age']<30]
print('young people:', young_people[:5])
```

```python
#Access records with indices 0,10,20

selected_records = data[[0,10,20]]
print('Selected Records:\n', selected_records)
```

```python
#Access the first name and city columns for all records
cities_and_firstnames = data[['City', 'FirstName']]
print("Cities and FirstNames:\n", cities_and_firstnames)
```

Broadcasting

Basic operations on numpy arrays (addition, etc.) are elementwise

This works on arrays of the same size. Nevertheless, It’s also possible to do operations on arrays of different sizes if NumPy can transform these arrays so that they all have the same size: this conversion is called broadcasting.

 image.png

```python
a = np.tile(np.arange(0,40,10), (3,1))
print(a)

print("$$$$$$$$$$")
a = a.T
print(a)
```

```python
b = np.array([0,1,2])
b
```

```python
a + b
```

```python
# Attempting to broadcast arrays that cannot be broadcast according to the rules will result in an error:
a = np.array([1, 2, 3])
b = np.array([1, 2])  # Different shape

# This will raise a ValueError due to broadcasting rules
try:
    result = a + b
    print(result)
except ValueError as e:
    print(e)
```
